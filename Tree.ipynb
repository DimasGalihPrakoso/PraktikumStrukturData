```python
class Node:
    def __init__(self, data):
      self.data = data
      self.left = None
      self.right = None

#Membuat tree secara manual
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

#Fungsi inorder traversal
def inorder(node):
    if node:
        inorder(node.left)
        print(node.data, end=' ')
        inorder(node.right)

print("Hasil inorder traversal dari binary tree:")
inorder(root)
```

```python
class Node:
      def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

class BinarySearchTree:
  def __init__(self):
    self.root = None
  def insert(self, root, data):
    if root is None:
      return Node(data)
    if data < root.data:
      root.left = self.insert(root.left, data)
    else:
      root.right = self.insert(root.right, data)
    return root

bst = BinarySearchTree()
root = None
data_list = [50, 30, 70, 20, 40, 60, 80]
for value in data_list:
  root = bst.insert(root, value)
```

```python
def inorder(node):
  if node:
    inorder(node.left)
    print(node.data, end=' ')
    inorder(node.right)

def preorder(node):
  if node:
    print(node.data, end=' ')
    preorder(node.left)
    preorder(node.right)

def postorder(node):
  if node:
    postorder(node.left)
    postorder(node.right)
    print(node.data, end=' ')

print("Inorder Traversal:")
inorder(root)
print("\nPreorder Traversal:")
preorder(root)
print("\nPostorder Traversal:")
postorder(root)
```

```python
#Tugas 1
#Program Struktur Binary Tree Manual

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


def hitung_jumlah_huruf(nama):
    """Menghitung jumlah huruf (tanpa spasi) pada nama."""
    return sum(1 for ch in nama if ch.isalpha())


def hitung_jumlah_vokal(nama):
    """Menghitung jumlah huruf vokal (a, i, u, e, o) pada nama."""
    vokal = "aiueo"
    return sum(1 for ch in nama.lower() if ch in vokal)


def inorder(node, hasil):
    """Traversal inorder: kiri -> root -> kanan."""
    if node is not None:
        inorder(node.left, hasil)
        hasil.append(node.data)
        inorder(node.right, hasil)


def tampilkan_struktur(node, level=0, label="Root"):
    """Menampilkan struktur tree secara visual (pre-order, indentasi per level)."""
    if node is not None:
        print("    " * level + f"{label}: {node.data}")
        tampilkan_struktur(node.left, level + 1, "Kiri")
        tampilkan_struktur(node.right, level + 1, "Kanan")


nama_lengkap = "Dimas Galih Prakoso"
npm = "2505060023"
tanggal_lahir = "12-05-2007"  #format DD-MM-YYYY

#Pengolahan data
jumlah_huruf = hitung_jumlah_huruf(nama_lengkap)      #Node root
jumlah_vokal = hitung_jumlah_vokal(nama_lengkap)      #Anak kiri root
dua_digit_terakhir_npm = int(npm[-2:])                #Anak kanan root

dd, mm, yyyy = tanggal_lahir.split("-")
tanggal = int(dd)   #Anak kiri dari node kiri
bulan = int(mm)     #Anak kanan dari node kanan

root = Node(jumlah_huruf)              #Root: jumlah huruf nama
root.left = Node(jumlah_vokal)         #Anak kiri root: jumlah huruf vokal
root.right = Node(dua_digit_terakhir_npm)  #Anak kanan root: 2 digit terakhir NPM

root.left.left = Node(tanggal)         #Anak kiri dari node kiri: tanggal lahir (DD)
root.right.right = Node(bulan)         #Anak kanan dari node kanan: bulan lahir (MM)

print("=" * 50)
print("IDENTITAS:")
print(f"Nama  : {nama_lengkap}")
print(f"NPM   : {npm}")
print(f"TTL   : {tanggal_lahir}")
print("=" * 50)

print("\nStruktur Tree (Pre-order dengan indentasi):")
tampilkan_struktur(root)

print("\nDetail Node:")
print(f"Root (jumlah huruf nama)             : {root.data}")
print(f"Anak Kiri Root (jumlah vokal)        : {root.left.data}")
print(f"Anak Kanan Root (2 digit NPM)        : {root.right.data}")
print(f"Anak Kiri dari Kiri (tanggal lahir)  : {root.left.left.data}")
print(f"Anak Kanan dari Kanan (bulan lahir)  : {root.right.right.data}")

hasil_inorder = []
inorder(root, hasil_inorder)
print("\nHasil Inorder Traversal:")
print(" -> ".join(str(x) for x in hasil_inorder))
```

```python
#tugas 2
#Program Binary Search Tree (BST)

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def insert(root, data):
    if root is None:
        return Node(data)
    if data < root.data:
        root.left = insert(root.left, data)
    elif data > root.data:
        root.right = insert(root.right, data)
    return root

def inorder(root, hasil=[]):
    if root:
        inorder(root.left, hasil)
        hasil.append(root.data)
        inorder(root.right, hasil)
    return hasil

def preorder(root, hasil=[]):
    if root:
        hasil.append(root.data)
        preorder(root.left, hasil)
        preorder(root.right, hasil)
    return hasil

def postorder(root, hasil=[]):
    if root:
        postorder(root.left, hasil)
        postorder(root.right, hasil)
        hasil.append(root.data)
    return hasil

def search(root, target):
    if root is None:
        return False
    if root.data == target:
        return True
    return search(root.left, target) if target < root.data else search(root.right, target)


#Data angka: 3 digit terakhir NPM, tanggal, bulan, ASCII 2 huruf nama, 2 angka bebas
angka = [23, 12, 5, 68, 105, 7, 99]

root = None
for a in angka:
    root = insert(root, a)

print("Inorder  :", inorder(root, []))
print("Preorder :", preorder(root, []))
print("Postorder:", postorder(root, []))

#Pencarian
print("\nCari 23:", "Ditemukan" if search(root, 23) else "Tidak ditemukan")
print("Cari 50:", "Ditemukan" if search(root, 50) else "Tidak ditemukan")
```
