```python
#TUgas
from collections import deque

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D', 'E'],
    'D': ['B', 'C', 'E'],
    'E': ['C', 'D']
}
# Sisi: A-B, A-C, B-C, B-D, C-D, C-E, D-E = 7 sisi, 5 simpul

def print_graph(graph):
    """Menampilkan setiap simpul beserta daftar tetangganya."""
    for node in graph:
        print(f"{node} -> {graph[node]}")

print("=== Tugas 1: Graf Baru (5 simpul, 7 sisi) ===")
print_graph(graph)

def bfs(graph, start):
    """BFS: menjelajahi graf level demi level menggunakan queue."""
    visited = set()
    queue = deque([start])
    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            print(vertex, end=" ")
            visited.add(vertex)
            queue.extend([n for n in graph[vertex] if n not in visited])

def dfs(graph, start, visited=None):
    """DFS: menjelajahi graf secara rekursif, menyelam ke satu cabang dahulu."""
    if visited is None:
        visited = set()
    if start not in visited:
        print(start, end=" ")
        visited.add(start)
        for neighbor in graph[start]:
            dfs(graph, neighbor, visited)

print("\n=== Tugas 2: BFS dan DFS dari simpul 'A' ===")
print("BFS: ", end="")
bfs(graph, 'A')
print("\nDFS: ", end="")
dfs(graph, 'A')

def bfs_list(graph, start):
    """Versi BFS yang mengembalikan list urutan kunjungan, bukan mencetak langsung."""
    visited = set()
    queue = deque([start])
    urutan = []
    while queue:
        vertex = queue.popleft()
        if vertex not in visited:
            urutan.append(vertex)
            visited.add(vertex)
            queue.extend([n for n in graph[vertex] if n not in visited])
    return urutan

print("\n\n=== Tugas 3: BFS Mengembalikan List Urutan Kunjungan ===")
hasil_bfs_list = bfs_list(graph, 'A')
print(f"Urutan kunjungan BFS dari 'A': {hasil_bfs_list}")

def find_path(graph, start, end, path=None):
    """Mencari satu jalur dari start ke end menggunakan DFS rekursif."""
    if path is None:
        path = []
    path = path + [start]
    if start == end:
        return path
    for neighbor in graph[start]:
        if neighbor not in path:
            hasil = find_path(graph, neighbor, end, path)
            if hasil:
                return hasil
    return None

print("\n=== Tugas 4: Mencari Jalur dengan find_path() ===")
jalur = find_path(graph, 'A', 'E')
print(f"Jalur dari 'A' ke 'E': {jalur}")

def is_connected(graph):
    """Mengecek konektivitas graf: jalankan DFS dari satu simpul,
    lalu bandingkan jumlah simpul terkunjungi dengan jumlah simpul total."""
    simpul_awal = next(iter(graph))
    visited = set()

    def _dfs(node):
        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                _dfs(neighbor)

    _dfs(simpul_awal)
    return len(visited) == len(graph)

print("\n=== Tugas 5: Mengecek Konektivitas Graf dengan is_connected() ===")
print(f"Apakah graf di atas terhubung seluruhnya? {is_connected(graph)}")

# Uji tambahan: graf yang sengaja dibuat tidak terhubung (dua komponen terpisah)
graf_tidak_terhubung = {
    'A': ['B'],
    'B': ['A'],
    'C': ['D'],
    'D': ['C']
}
print(f"Apakah graf terpisah (A-B, C-D) terhubung seluruhnya? {is_connected(graf_tidak_terhubung)}")
```
